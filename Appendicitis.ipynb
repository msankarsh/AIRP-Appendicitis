# Quick TP to Req Columns


```python
image_dir = "US_Pictures"
metadata_file = "app_data_clean.csv"


required_columns = [
    'Appendix_Diameter', 'WBC_Count','BMI','Sex','Loss_of_Appetite','Body_Temperature',
    # Add more features here you would use
]
```

```python
pattern = r"^(\d+)\.(\d+)\s*(.*?)\.bmp$"
```

```python
os.listdir("US_Pictures")
```

```python
os.path.join("US_Pictures", "train", "1.2 Ileitis.png")
```

```python
image_records = []
for filename in os.listdir(image_dir):

    # Only process bmp files
    if not filename.lower().endswith(".bmp"):
        continue

    filepath = os.path.join(image_dir, filename)

    # Expected format:
    # {pid}.{im_number} {description}.bmp
    #
    # Example:
    # 123.4 Abdomen transverse.bmp

    match = re.match(
        r"^(\d+)\.(\d+)\s*(.*?)\.bmp$",
        filename,
        flags=re.IGNORECASE
    )

    if match:
        pid = int(match.group(1))
        image_number = int(match.group(2))
        description = match.group(3).strip()

    else:
        # Handle unexpected filenames
        # Try to at least recover the patient ID
        fallback = re.match(r"^(\d+)", filename)

        if fallback:
            pid = int(fallback.group(1))
        else:
            pid = None

        image_number = None
        description = filename.replace(".bmp", "")

    image_records.append({
        "pid": pid,
        "image_number": image_number,
        "description": description,
        "filename": filename
    })

images_df = pd.DataFrame(image_records)
```

```python
images_df.head()
```

```python
len(images_df)
```

```python
image_summary = (
        images_df
        .groupby("pid")
        .agg(
            num_images=("filename", "count"),
            image_numbers=("image_number", lambda x: sorted(
                [i for i in x.dropna().unique()]
            )),
            image_descriptions=("description", lambda x: sorted(
                list(set(x))
            ))
        )
        .reset_index()
    )
```

```python
image_summary.head()
```

```python
metadata = pd.read_csv(metadata_file)


# Convert US_Number from float to integer where possible
metadata["US_Number_clean"] = (
    pd.to_numeric(metadata["US_Number"], errors="coerce")
    .dropna()
    .astype(int)
)

"""
errors = 'coerce' tells pandas, "if you find a value that cannot be converted, replace it with missing
(NaN) instead of throwing an error"
"""

# Unique patients appearing in metadata
metadata_patients = (
    metadata["US_Number_clean"]
    .dropna()
    .unique()
)
```

```python
patients_without_images = sorted(
    set(metadata_patients) -
    set(image_summary["pid"])
)
```

```python
len(patients_without_images)
```

```python
# All unique patients from either source
all_patients = pd.DataFrame({
    "pid": sorted(
        set(image_summary["pid"])
        .union(set(metadata_patients))
    )
})
```

```python
final_df = (
    all_patients
    .merge(
        image_summary,
        on="pid",
        how="left"
    )
)


# Add image presence indicator
final_df["has_images"] = (
    final_df["num_images"]
    .fillna(0)
    .astype(int)
    .gt(0) # Greater than 0
)

final_df["num_images"] = (
    final_df["num_images"]
    .fillna(0)
    .astype(int)
)
```

```python
# Does patient appear in metadata?
metadata_pid_set = set(metadata_patients)

final_df["appears_in_metadata"] = (
    final_df["pid"]
    .isin(metadata_pid_set)
)
```

```python
# Add metadata completeness information

# Keep only the needed columns plus patient identifier
metadata_subset = metadata[["US_Number_clean"] + required_columns].copy()

metadata_subset = (
        metadata_subset
        .rename(columns={"US_Number_clean": "pid"})
        .drop_duplicates("pid")
    )

final_df = final_df.merge(
        metadata_subset,
        on="pid",
        how="left"
    )

final_df["all_required_fields_present"] = (
        final_df[required_columns]
        .notna()
        .all(axis=1)
    )
```

```python
final_df = final_df.sort_values("pid").reset_index(drop=True)
print(len(final_df)) # number of patients we have either in metadata or in images folder
final_df.head(100)
```
